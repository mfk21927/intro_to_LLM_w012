```python
from transformers import pipeline

# Load GPT-2
generator = pipeline('text-generation', model='gpt2')

prompts = [
    "The future of artificial intelligence in software development is",
    "Deep learning models are essentially"
]

gen_results = []
print("--- Generating Text ---")
for prompt in prompts:
    # Experimenting with temperature and top_k
    output = generator(prompt, max_length=50, num_return_sequences=1, 
                       temperature=0.8, top_k=50, truncation=True)
    text = output[0]['generated_text']
    gen_results.append(f"Prompt: {prompt}\nGenerated: {text}\n")
    print(text)

# Save to file
with open("results_generation.txt", "w") as f:
    f.writelines(gen_results)
print("\nResults saved to results_generation.txt")
```

```python
from transformers import pipeline

# Load BERT
filler = pipeline('fill-mask', model='bert-base-uncased')

mask_prompt = "The programmer decided to [MASK] the code to fix the bug."
fill_results = filler(mask_prompt)

mask_output = f"Prompt: {mask_prompt}\n"
print("--- Fill-Mask Predictions ---")
for res in fill_results:
    line = f"Score: {res['score']:.4f} | Prediction: {res['token_str']}"
    mask_output += line + "\n"
    print(line)

# Append/Save to file
with open("results_mask_qa.txt", "w") as f:
    f.write("--- MASK TASK ---\n" + mask_output + "\n")
print("\nResults saved to results_mask_qa.txt")
```

```python
from transformers import pipeline

# Load DistilBERT for QA
qa_model = pipeline('question-answering', model='distilbert-base-cased-distilled-squad')

context = """
Hugging Face is a technology company based in New York and Paris. 
They are best known for the Transformers library, which allows developers 
to easily download and train state-of-the-art machine learning models.
"""
question = "Where is Hugging Face based?"

qa_res = qa_model(question=question, context=context)
qa_output = f"Context: {context}\nQuestion: {question}\nAnswer: {qa_res['answer']}\nScore: {qa_res['score']:.4f}"

print(f"--- QA Task ---\nAnswer: {qa_res['answer']}")

# Append to the mask results file
with open("results_mask_qa.txt", "a") as f:
    f.write("\n--- QA TASK ---\n" + qa_output)
print("\nResults appended to results_mask_qa.txt")
```
